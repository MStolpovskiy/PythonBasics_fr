## Ex. 2 : Apprendre le bon style

L'objectif de cet exercice est de corriger le code suivant. Ce code affiche une construction mathématique particulière : un nombre complexe. Pour l'instant, peu importe comment il fait cela. Copiez simplement le code dans la cellule ci-dessous et modifiez-le selon les instructions suivantes :

```python
    def printComplexNumber(realpartofcomplexnumber = 0,imagpartofcomplexnumber= 0 ) :
        R=   realpartofcomplexnumber
        I=imagpartofcomplexnumber
        return f"({R } + { I}i)"
    Longvariablename=printComplexNumber(3.0,-4.5)
    print( Longvariablename)
```

Règles de bon style en Python (liste non exhaustive) :

- Les opérateurs `=`, `+`, `-`, etc. sont entourés d'espaces
- Les variables, les arguments de fonction et les fonctions sont nommés en minuscules, et les noms longs sont séparés par `_`
  - dans le code ci-dessus, `printComplexNumber` est un nom de fonction
  - `R`, `I`, `Longvariablename` sont des variables
  - `realpartofcomplexnumber` et `imagpartofcomplexnumber` sont des arguments de fonction
- Le nom de la variable ou de la fonction doit permettre de comprendre globalement ce qu'elle fait
- La définition de fonction (dans notre exemple, il s'agit de la ligne avec `def` et des trois lignes en dessous) est séparée du reste du code par des lignes vides
- Le bloc d'instructions `import` est séparé par une ligne vide du reste du code (ce qui se trouve sous `import`)
- Les parties logiquement distinctes du code sont séparées par une ligne vide (par exemple, après la définition d'une fonction, ajoutez une ligne vide)
- Les commentaires commencent après le symbole `#`. Tous les passages peu évidents du code doivent être commentés.
- La virgule est toujours suivie d'une espace.
- Les espaces à côté des parenthèses sont inutiles. Par exemple, `(x)` est correct. `( x )` est faux (bien que fonctionnel).
- L'opérateur point n'est pas entouré d'espaces. Les opérateurs `**` et `%` non plus.
- Les lignes longues doivent être coupées. Après une virgule, vous pouvez passer à la ligne en toute sécurité. Il faut aligner la suite pour relier visuellement les lignes. S'il n'y a pas de virgule, vous pouvez utiliser le caractère de continuation de ligne `\`
- Pensez à ajouter de la documentation à toute fonction :

```python
    def fun():
        '''
        Fonction qui ne fait absolument rien
        '''
        pass
```

## Ex. 3 : Utiliser Python comme une calculatrice avancée

Dans cet exercice, vous allez utiliser Python pour effectuer des calculs mathématiques variés : manipuler des nombres irrationnels, travailler avec les nombres complexes et observer comment Python gère les très grands nombres. Cet exercice a 3 parties, voici la premiere :

### 1. Calcul algébrique avec une racine carrée

Déclarez la variable $x = \frac{1 + \sqrt{5}}{2}$, puis calculez $x^2 - x$ et affichez le résultat.

* **Puissance :** pour élever un nombre à une puissance, vous pouvez écrire `x * x` ou, de manière plus concise et lisible, utiliser l'opérateur `**` (par exemple `x**2` pour `x` en carée).
* **Racine carrée :** utilisez la puissance fractionnaire : prendre la racine carrée d'un nombre revient à l'élever à la puissance `0.5` (par exemple `9**0.5` c'est racine carée de 9, alors 3).
* **Priorité des opérations :** pensez impérativement aux parenthèses autour du numérateur `(1 + 5**0.5)` avant de diviser par `2`, afin que la division s'applique à l'ensemble du numérateur et non uniquement à la racine carrée.
* **Astuce mathématique :** vous pouvez anticiper et vérifier votre résultat ! $x = \frac{1 + \sqrt{5}}{2}$ - c'est le nombre d'or. Par définition du nombre d'or, $x$ vérifie l'équation $x^2 - x = 1$. On s'attend donc à obtenir une valeur égale (ou extrêmement proche, à cause de la représentation en virgule flottante `float`) à `1`.
* **Remarque sur les flottants :** les nombres à virgule en Python (`float`) sont stockés en base 2 avec une précision finie ; il est donc fréquent d'observer de minuscules écarts d'arrondi lors de calculs algébriques.

### 2. Calcul avec les nombres complexes

Déclarez la variable $z = i$ (l'unité imaginaire, telle que $i^2 = -1$), puis calculez $z^2$ et affichez le résultat.

* **Notation des complexes en Python :** en mathématiques, on note l'unité imaginaire $i$, mais en Python on utilise la lettre `j` collée directement au nombre (par exemple `1j` pour $i$, ou `3j` pour $3i$). N'écrivez pas `j` seul, sinon Python cherchera une variable nommée `j` et lèvera une erreur si elle n'a pas été définie.
* **Exemple de nombre complexe :** on déclare un nombre complexe de la forme $a + bi$ en écrivant par exemple `n = 2 - 3j`. Python crée alors un objet de type `complex` composé d'une partie réelle et d'une partie imaginaire.
* **Calcul de $z^2$ :** une fois `z = 1j` défini, utilisez `z**2` pour vérifier que Python retrouve bien `-1` (ou `(-1+0j)` selon le contexte d'affichage). Notez que l'opération de puissance fonctionne directement sur les nombres complexes sans syntaxe particulière.

### 3. Grands entiers et dépassement de capacité (*overflow*)

Calculez et affichez la valeur de $2^{1000}$ en Python.

* **Le piège des autres langages :** dans de nombreux langages de programmation (comme le C, le C++ ou Java), les types entiers natifs occupent un nombre fixe de bits en mémoire (par exemple 32 bits ou 64 bits), ce qui borne la valeur maximale représentable. Calculer $2^{1000}$ dans un tel entier provoque un dépassement de capacité (*integer overflow*) : le calcul échoue, déclenche une erreur ou produit un résultat tronqué / négatif incohérent.
* **La particularité de Python :** en Python, le type entier (`int`) possède une précision arbitraire : il n'a pas de limite fixe de taille imposée par le type de base. En principe, un entier peut croître tant que la mémoire disponible de l'ordinateur le permet. Vous obtiendrez donc un nombre entier exact de plus de 300 chiffres décimaux sans aucune erreur de dépassement.

## Opérateurs d'affectation augmentée (`+=`, `-=`, `*=`, `/=`)

Python propose des raccourcis pratiques appelés **opérateurs d'affectation augmentée** pour rendre ces mises à jour plus concises et plus lisibles.

## Conversion de types (*type casting*)

En Python, il arrive souvent que l'on doive changer le type d'une donnée : par exemple, transformer un nombre en texte pour l'afficher dans une phrase, ou convertir une chaîne de caractères saisie par l'utilisateur en nombre pour effectuer un calcul. Python fournit des fonctions intégrées comme `str()`, `int()` et `float()` pour réaliser ces conversions.

---

### 1. Les principales fonctions de conversion

Testez chacune de ces conversions dans Python et observez le résultat obtenu ainsi que son type (avec `type(...)`) :

- **Convertir un nombre en chaîne de caractères (`str(...)`) :** par exemple `str(5)` renvoie `'5'`. Même si le contenu ressemble à un chiffre, il s'agit désormais d'une chaîne (`str`) : vous pouvez par exemple la concaténer avec d'autres textes (`"Valeur : " + str(5)`), alors que `"Valeur : " + 5` provoquerait une erreur `TypeError`.
- **Convertir une chaîne en entier (`int(...)`) :** par exemple `int('5')` renvoie l'entier `5` (de type `int`), sur lequel vous pouvez ensuite faire des calculs mathématiques comme `int('5') + 3` (`8`). Attention : la chaîne doit représenter un entier valide (par exemple `int('bonjour')` ou `int('5.2')` lèvera une erreur `ValueError`).
- **Convertir un entier en flottant (`float(...)`) :** par exemple `float(5)` renvoie `5.0` (de type `float`). Cela permet par exemple de forcer un calcul décimal ou d'harmoniser le type d'une variable numérique.

---

### 2. Mise en pratique et cas limites

**À faire :**

- **Utiliser une conversion dans un calcul ou un affichage :** par exemple, essayez d'afficher `"J'ai " + str(16) + " ans"` ou de calculer `int('12') + float('3.5')`.
- **Observer les limites de conversion :** testez `int('abc')` ou `float('3.14')` et notez ce qui fonctionne et ce qui déclenche une erreur (`ValueError`). Remarquez aussi ce que donne `int(3.9)` : la conversion d'un flottant vers un entier tronque la partie décimale (elle ne fait pas un arrondi au plus proche).

### 1. Guillemets simples, doubles et inclusion d'un guillemet dans le texte

En Python, une chaîne de caractères peut être délimitée de deux façons :

- **Guillemets simples (`'...'`) ou guillemets doubles (`"..."`) :** les deux sont équivalents pour créer une chaîne (par exemple `'bonjour'` et `"bonjour"`). Choisir l'un ou l'autre permet souvent d'éviter d'avoir à échapper certains caractères.
- **Inclure un guillemet dans une chaîne :** comment écrire une chaîne qui contient une apostrophe ou un guillemet (par exemple `It's ok` ou `Il a dit "Bonjour"`) ? Vous avez deux solutions :
  - Utiliser l'autre type de guillemets pour encadrer la chaîne : par exemple `"It's ok"` (guillemets doubles autour d'une apostrophe simple) ou `'Il a dit "Bonjour"'` (guillemets simples autour de guillemets doubles).

### 2. Déclarer proprement des chaînes longues ou multi-lignes

Lorsque votre texte est long ou s'étale sur plusieurs lignes, écrire une seule ligne interminable nuit à la lisibilité (rappel de style : les lignes longues doivent être coupées proprement).

- **Concaténation implicite entre parenthèses :** en Python, plusieurs chaînes littérales placées côte à côte à l'intérieur de parenthèses sont automatiquement collées par l'interpréteur, même sur plusieurs lignes :

```python
        s = (
            'very '
            'long '
            'string'
        )
```

  Pensez bien aux espaces en fin ou en début de chaque sous-chaîne (ici `'very '` et `'long '`), sinon les mots seront collés (`'verylongstring'`).

- **Chaînes multi-lignes avec triples guillemets (`'''...'''` ou `"""..."""`) :** elles permettent d'écrire un texte sur plusieurs lignes en conservant les retours à la ligne :

```python
        s = '''
        A very
        very
        long
        string
        '''
```

  Attention : un retour à la ligne juste après `'''` ou avant `'''` fait partie de la chaîne (`\n`), et l'indentation placée à l'intérieur des triples guillemets est également conservée telle quelle dans le texte.

---

### 3. Caractères spéciaux (`\n`, `\t`, `\\`) et chaînes brutes (*raw strings*)

Certaines séquences commençant par l'antislash `\` ont une signification particulière dans une chaîne classique :

- **Caractères spéciaux `\n` et `\t` :** `\n` insère un **saut de ligne** (*newline*) et `\t` insère une **tabulation** (*tab*). Testez par exemple `print("Ligne 1\nLigne 2")` et `print("Col1\tCol2")`.
- **Afficher le symbole `\` lui-même (`\\`) :** comme `\` sert à introduire des séquences spéciales, pour représenter un seul caractère antislash dans une chaîne standard, il faut le doubler : `'\\'`. Testez `print('Chemin : C:\\Users\\Eleve')`.
- **Chaînes brutes (*raw string*, préfixées par `r` ou `R`) :** avec `r'This is the raw string \n'`, Python n'interprète pas `\n` comme un saut de ligne : les caractères `\` et `n` restent littéraux. C'est très pratique pour des expressions régulières ou des chemins de fichiers Windows, tout en gardant à l'esprit qu'une *raw string* ne peut pas se terminer par un `\` isolé non échappé.

---

### 4. Accéder aux caractères par leur position (indexation) et longueur (`len`)

Chaque caractère d'une chaîne possède une position appelée **indice** (*index*) :

- **Comptage à partir de `0` (toujours !) :** dans une chaîne `s`, le premier caractère est à l'indice `0`, le deuxième à l'indice `1`, le troisième à l'indice `2`, le quatrième à l'indice `3`, etc.

```python
        s = 'abcdefg'
        print(s[3])  # affiche le 4e caractère ('d'), car on compte à partir de 0 !
```

  Vérifiez : `s[0]` vaut `'a'`, `s[1]` vaut `'b'`, `s[2]` vaut `'c'` et `s[3]` vaut `'d'`. (En Python, on peut aussi utiliser des indices négatifs comme `s[-1]` pour le dernier caractère, ou lever une erreur `IndexError` si l'indice dépasse la taille de la chaîne.)

- **Longueur d'une chaîne avec `len(s)` :** la fonction intégrée `len(s)` renvoie le nombre total de caractères contenus dans la chaîne `s` (espaces et caractères spéciaux compris). Testez `len('abcdefg')` puis `len('Hello World')`.

---

### 5. Découper une chaîne avec `split()`

La méthode `.split()` permet de découper une chaîne de caractères en une **liste** de sous-chaînes selon un séparateur :

```python
        s = 'a string to be splitted'
        print(s.split()[4])
        print(s.split('t')[2])
```

- **Avec un séparateur précis (`s.split('t')`) :** découpe la chaîne à chaque occurrence du caractère `'t'` (le `'t'` lui-même est retiré). Observez la liste obtenue par `s.split('t')` et vérifiez quel est l'élément d'indice `2`.

## Chaînes formatées (*f-strings*)

Il arrive très souvent que l'on doive afficher un message qui combine du texte fixe et des valeurs stockées dans des variables, par exemple :

    "Google shares rise 5.25%"

Dans cet exemple, `"Google"`, `"rise"` (ou `"fall"`) et `5.25` sont déterminés dynamiquement à partir de variables. Plutôt que de coller des morceaux de texte avec `+` et de convertir manuellement chaque nombre avec `str(...)`, Python propose les **chaînes formatées** (appelées *f-strings*, préfixées par `f` ou `F`).

---

### 1. Principe des *f-strings*

Une *f-string* commence par la lettre `f` (ou `F`) juste avant le guillemet d'ouverture (`f"..."` ou `f'...'`). À l'intérieur, toute expression placée entre accolades `{...}` est évaluée par Python, convertie en chaîne de caractères et insérée directement dans le texte.

Testez l'exemple suivant :

    company_name = 'Google'
    growth = 5.253
    # la ligne suivante sera expliquée une fois les tableaux / listes abordés
    rise_or_fall = ['rise', 'fall'][int(growth < 0)]
    print(f"{company_name} shares {rise_or_fall} {abs(growth):.2f}%")

- **Insertion de variables :** `{company_name}` insère le contenu de la variable `company_name` (`Google`).
- **Choix automatique du mot (`rise` ou `fall`) :** `growth < 0` est une condition qui vaut `True` (soit `1` une fois convertie en entier avec `int(...)`) lorsque la variation est négative, et `False` (soit `0`) lorsqu'elle est positive ou nulle. On sélectionne ainsi l'élément d'indice `0` (`'rise'`) ou d'indice `1` (`'fall'`). Ce mécanisme sera détaillé lorsque vous étudierez les listes/tableaux.
- **Valeur absolue avec `abs(...)` :** comme le message indique déjà s'il s'agit d'une hausse (`rise`) ou d'une baisse (`fall`), on affiche `abs(growth)` afin d'avoir un pourcentage positif dans les deux cas.

---

### 2. Formater l'affichage des nombres (`:.2f` et autres options)

À l'intérieur des accolades d'une *f-string*, vous pouvez ajouter un deux-points `:` suivi d'une spécification de format pour contrôler l'apparence d'une valeur :

- **Limiter le nombre de décimales avec `:.2f` :** dans l'exemple, `growth` vaut `5.253`, mais `{abs(growth):.2f}` affiche `5.25`. Le `f` indique un format décimal à virgule fixe (*fixed-point*), et `.2` demande **2 chiffres après la virgule** (le point décimal en Python), avec arrondi automatique.
- **Autres formats numériques :** il existe de nombreuses autres options de formatage en Python (par exemple pour afficher un pourcentage avec `:%`, forcer un signe `+` ou `-` avec `:+`, aligner un texte ou ajouter des zéros en tête avec `:05d`, ou écrire en notation scientifique avec `:.3e`).

Oui, `---...---` correspond bien au signal de détresse **SOS** en code Morse (trois traits `---` pour **S**, trois points `...` pour **O**, et trois traits `---` pour **S**).

---

### 1. Construire le signal SOS sans l'écrire en entier

En code Morse, le signal international de détresse **SOS** s'écrit `---...---` :
- la lettre **S** est codée par trois traits (`---`),
- la lettre **O** est codée par trois points (`...`),
- la lettre **S** est de nouveau codée par trois traits (`---`).

**À faire :** affichez 10 fois le signal SOS (`"---...---"`), sans écrire la chaîne littérale `"---...---"` directement dans votre code.

* **Construire chaque lettre :** utilisez l'opérateur de répétition `*` sur un seul caractère :

        lettre_s = '-' * 3
        lettre_o = '.' * 3

* **Assembler le signal SOS :** utilisez la concaténation `+` pour former le signal complet :

        sos = lettre_s + lettre_o + lettre_s

  Vérifiez avec `print(sos)` que vous obtenez bien `'---...---'`, sans jamais avoir écrit cette chaîne complète dans votre code source.

---

### 2. Répéter le signal SOS 10 fois

Une fois le signal construit dans la variable `sos`, multipliez-le par `10` avec l'opérateur `*` pour l'afficher 10 fois de suite :

    print(sos * 10)

* **Comportement de `*` sur une chaîne :** multiplier une chaîne par l'entier `10` répète cette chaîne 10 fois à la suite sur la même ligne.
* **Variante en une seule expression :** vous pouvez aussi écrire le calcul directement dans `print(...)` en utilisant des parenthèses pour regrouper le signal SOS avant de le répéter 10 fois :

        print(('-' * 3 + '.' * 3 + '-' * 3) * 10)

* **Attention aux priorités des opérateurs :** en Python, `*` est prioritaire sur `+`. Sans les parenthèses autour de `'-' * 3 + '.' * 3 + '-' * 3`, le `* 10` ne s'appliquerait qu'au dernier `'-' * 3`. Les parenthèses garantissent que c'est bien tout le signal SOS qui est répété 10 fois.

## Ex. 5 : Logique booléenne

### 1. Valeurs booléennes et branchement conditionnel (`if`)

La deuxième ligne n'est exécutée que si `my_salary` est strictement inférieur à un million. Attention à l'indentation (4 espaces en Python) : elle délimite le bloc exécuté lorsque la condition est vraie.

Seul le premier bloc dont la condition est vraie est exécuté ; `else` capture tous les autres cas.
